# ENTRAINEMENT A LA VISUALISATION AVEC SEABORN

```python
import seaborn as sns
```

Les fonctions de seaborn ont presque toutes la meme struture : 

**sns.fonction(x, y, data, hue, size, style)**

- x, y, data = Données a afficher
- hue, size, style = options de segmentation

```python
titanic = sns.load_dataset('titanic')
titanic.head()
```

```python
titanic = titanic.drop(['who', 'adult_male', 'alone', 'alive', 'embark_town', 'class'], axis=1)
titanic.dropna(axis=0, inplace=True)
titanic.head()
```

## Viz des relations 

```python
sns.pairplot(titanic)
```

## Viz de l'age des passagers suivant la classe

```python
sns.catplot(x='pclass', y='age', data=titanic)
```

#### Suivant le sex

```python
sns.catplot(x='pclass', y='age', data=titanic, hue='sex')
```

## Viz avec boxplot

```python
sns.boxplot(x='pclass', y='age', data=titanic, hue='sex')
```

## Viz de la distribution

```python
# Distrbution du tarif
sns.distplot(titanic['fare'])
```

## Distribution suivant deux variables

```python
sns.jointplot(x = 'age', y = 'fare', data=titanic)
```

```python
# Suivant le Kernel Density Estimation
sns.jointplot(x = 'age', y = 'fare', data=titanic, kind='kde')
```

```python
# Style hexagonal
sns.jointplot(x = 'age', y = 'fare', data=titanic, kind='hex')
```

## Viz avec heatmap

```python
sns.heatmap(titanic.corr(numeric_only=True))
```

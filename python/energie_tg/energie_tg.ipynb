**Contexte**: Le secteur de l'énergie au Togo repose sur plusieurs sources : le thermique, l'hydraulique (notamment le barrage de Nangbéto), le solaire (en développement rapide, notamment avec la centrale de Blitta), l'éolien et la biomasse. Plusieurs fournisseurs interviennent sur le marché, dont la CEET (Compagnie Énergie Électrique du Togo) et la CEB (Communauté Électrique du Bénin). Tu travailles pour une structure chargée d'analyser des relevés de consommation et de production d'énergie collectés dans les différentes régions du pays (Maritime, Plateaux, Centrale, Kara, Savanes). Ces relevés remontent du terrain de façon peu homogène : erreurs de saisie, doublons, valeurs manquantes et incohérences de casse sont fréquents. ######

**Fichier fourni** : energie_togo.csv - relevés d'énergie par région/préfecture, avec les colonnes suivantes :

• id_releve : identifiant unique du relevé

• region, prefecture, localite : localisation géographique du relevé

• type_energie : Thermique, Hydraulique, Solaire, Eolienne ou Biomasse

• consommation_kwh : consommation relevée en kWh

• panne_signalee : Oui/Non, une panne a-t-elle été signalée lors du relevé

### NETTOYAGE

**Harmonisation des colonnes textuelles :**

```python
# Sur la région
energie['region'] = energie['region'].str.strip()
energie['region'] = energie['region'].str.title()

# Sur la préfecture
energie['prefecture'] = energie['prefecture'].str.strip()
energie['prefecture'] = energie['prefecture'].str.title()

# Sur la localité
energie['localite'] = energie['localite'].str.strip()
energie['localite'] = energie['localite'].str.title()

# Sur les fournisseurs
energie['fournisseur'] = energie['fournisseur'].str.strip()
energie['fournisseur'] = energie['fournisseur'].str.upper()

# Sur le type d’énergie
energie['type_energie'] = energie['type_energie'].str.strip()
energie['type_energie'] = energie['type_energie'].str.title()
```

```python
print("✅ Verification visuelle")
print()
textuel = ['region', 'prefecture', 'localite', 'fournisseur', 'type_energie']
for text in textuel :
    print(f"{text} : {energie[text].unique()}")
    print()
```

```python
#Suppression des doublons
energie = energie.drop_duplicates()
```

#### 1. Suppression des lignes
- Les lignes où la colonne **`fournisseur`**,  **`taux_couverture`** est vide (valeur manquante ou chaîne vide) sont supprimées.  
- Cela garantit que seules les observations avec un fournisseur valide sont conservées.

#### 2. Remplacement des valeurs manquantes
- Pour les colonnes **`consommation_kwh`** et **`production_kwh`**, les valeurs manquantes sont remplacées par la **médiane calculée par groupe de localité**.  
- Cette approche permet de conserver la cohérence des données tout en évitant de perdre des lignes entières.  
- La médiane est utilisée car elle est robuste face aux valeurs aberrantes.

```python
# Supprimer toutes les lignes où 'fournisseur et taux_couverture_pourcent ' est vide ou NaN
energie = energie.dropna(subset=['fournisseur', 'taux_couverture_pourcent'])
```

```python
# Remplacement des NaN par la médiane de chaque localité
for col in ['consommation_kwh', 'production_kwh']:
    energie[col] = energie[col].fillna(
        energie.groupby('localite')[col].transform('median')
    )
```

### Harmonisation de la date

```python
# Conversion en datetime
energie['date_releve'] = pd.to_datetime(energie['date_releve'], errors='coerce', dayfirst=True)
```

```python
print("Vérification de la conversion")
energie['date_releve'].isna().sum()
```

```python
# Supprimer les lignes où 'date_releve' est manquante
energie = energie.dropna(subset=['date_releve'])
```

```python
# Extraction de l'année et du mois
energie['annee'] = energie['date_releve'].dt.year
energie['mois'] = energie['date_releve'].dt.month
```

### CALCUL

```python
print()
reg_moy = energie.groupby('region').agg(
    conso_moy_kwh = ('consommation_kwh', 'mean'),
    taux_couv_moy_pct = ('taux_couverture_pourcent', 'mean'),
    prod_moy_kwh = ('production_kwh', 'mean')
)
print("Calcul de la moyenne de consommation, de couverture et de production par région : ")
reg_moy
```
